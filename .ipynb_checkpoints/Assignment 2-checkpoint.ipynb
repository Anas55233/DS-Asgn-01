```python
# Answer # 2

# Show a list which contain some duplicate items
shopping_list = ["apple", "banana", "apple", "orange", "banana", "grape"]

# Convert list into set data structure to remove duplicate items
shopping_set = set(shopping_list)
print(f"After converting into 'set': \n {shopping_set}")
print("Remove duplicate items successfully!")
print()

# Now convert into list again to show final output:
shopping_list = list(shopping_set)
print(f"Final Output in list: \n {shopping_list}")
```

#### Question No. 3

#### Answer No. 3

```python
# Answer # 3

# Firstly write course_a & course_b with their students:
course_a = {"John", "Alice", "Bob", "David"} 
course_b = {"Alice", "Eve", "Charlie", "David"}

# Now Find students who are in Course A but not in Course B
students_in_course_a_not_in_course_b = course_a.difference(course_b)
print(f"See students who are in Course A but not in Course B: \n\n{students_in_course_a_not_in_course_b}\n\n")


# Now Find items which are only one of both sets:
students_in_only_one_course = course_b.symmetric_difference(course_a)
print(f"See Students who are only in one of the two courses: \n\n |{students_in_only_one_course}")
```

#### Question No. 4

#### Answer No. 4

```python
# Answer # 4

# First make tuple of numbers:
numbers = (10, 20, 30, 40, 50, 60, 70)

# Secondly Convert into list from tuple
number_list = list(numbers)
print(f"The Number List below here: \n {number_list}")

# Display First three elements by slicing
first_3_elements = number_list[0:3]
print(f"\nShow First Three Elements: \n{first_3_elements}")

# Display Last three elements by slicing
last_3_elements = number_list[-3:]
print(f"\nShow Last Three Elements: \n{last_3_elements}")

# Display Alternate elements by slicing
alternate_elements = number_list[0::2]
print(f"\nShow Alternate Elements: \n{alternate_elements}")
```

#### Question No. 6

#### Answer No. 6

```python
# Answer # 6

# Make Two sets A and B:
set_A = {1, 2, 3, 4, 5}
set_B = {4, 5, 6, 7}

# First Find Common elements from both sets by using intersection
common_elements = set_A.intersection(set_B)
print(f"Common Elements from both sets are: \n {common_elements}")

# Now Find elements only in Set A.
common_elements_A = set_A.difference(set_B)
print(f"\nCommon Elements from set A are: \n {common_elements_A}")

# Now Find elements only in Set A.
common_elements_B = set_B.difference(set_A)
print(f"\nCommon Elements from set B are: \n {common_elements_B}")
```
