```python
# Answer # 1

# Declare variable "a" with value "15"
a = 15

# Declare variable "b" with value "4"
b = 4

#Now First make sum of variables
sum_variables = a + b

print("The sum of a and b is: ",sum_variables)
print()

# Now make difference from a to b
diff_variables = a - b

print("The Diffrence from a to b is : ", diff_variables)
print()

#Now declare product variable to multiply a into b
product_var = a * b

print("The product b/w a & b is: ", product_var)
print()

# Now make division after declare division_var
division_of_var = a / b

print("The result of a divided by b is: ", division_of_var)
print()

# Now find floor division of given variables:
floor_div_var = a // b

print("The result of floor division is: ", floor_div_var)
print()

# Now find modulus from a to b
var_modulus = a % b

print("The Modulus of a into b: ", var_modulus)
```

#### Question # 11

#### Answer # 11

```python
# Answer # 11

# Write variable product_code with thier value
product_code = "SKU-12345-XYZ"

# Check and print first position of dash(-)
first_dash_index = product_code.index("-")
print(f"The Index of first dash(-) is: '{first_dash_index}'")

# Now count and print number of dashes(-)
number_of_dashes = product_code.count("-")
print(f"The Count of dashes(-) are: '{number_of_dashes}'")

# Now Remove prefix (SKU-) and suffix (-XYZ) from product_code
product_code = product_code.removeprefix("SKU-")
product_code = product_code.removesuffix("-XYZ")

# Now print extract numeric part
print(f"The Updated Product Code after Extracting is '{product_code}'")
```

#### Question # 14

#### Answer # 14

```python
# Answer # 14

# First Write variable text with value:
text = "    Data Analysis with Python is very powerful    "
print(f"Original text: \n '{text}'")
print()

# First remove extra spaces using strip() and print:
strip_text = text.strip()
print(f"After Strip text: \n '{strip_text}'")
print()

# Convert the string into title case using title() and print:
title_text = strip_text.title()
print(f"After title text: \n '{title_text}'")
print()

# By using title_text, Now replace text "Powerful" with "Important".
replace_text = title_text.replace("Powerful", "Important")
print(f"After Replcing text: \n '{replace_text}'")
print()

# Now find and print position of word "Python":
index_python = title_text.index("Python")
print(f"The index of word 'Python' is: '{index_python}'")
print()
```

#### Question # 15

#### Answer # 15

```python
# Answer # 15

# First declare data list with values:
data = [12, 45, 2, 67, 34, 89, 23, 90]

print(f"Original data list: \n '{data}'")
print()

# Now Sort the list in ascending order and print same variable.
data.sort()
print(f"After Sorting data List: \n '{data}'")
print()

# Now Remove the highest value from the list and print updated list
data.remove(max(data))
print(f"After Removing max value from data list: \n '{data}'")
print()

# Now Insert the value 50 at index 3, and print list
data.insert(3, 50)
print(f"After insert value 50 at index 3 in data list: \n '{data}'")
print()

# Finally, reverse the updated list
data.reverse()
print(f"Finally, reverse Data list is: \n '{data}'")
print()

```
