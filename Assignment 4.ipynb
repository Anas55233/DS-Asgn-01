```python
# Ans No. 7

# Make a rondom number list:
num_list = [3, 5, 7, 8, 21, 15, 22, 26, 29, 31, 35]

# Make list comprehension using for loop that are divisible by 3 or 5:
divisible = [number for number in num_list if (number % 3 == 0) or (number % 5 == 0) ]

# print 
print(f"list containing only the numbers that are divisible by 3 or 5: \n{divisible}")
```

#### Answer No. 8

```python
# Ans No. 8

# Make list of numbers and multipliers
numbers = [2, 3, 4]
multipliers = [1, 2, 3]

print("Multiplication Table: \n")

# Make multiplication table by using nested loop:
for num in numbers:
    for multiply in multipliers:
        print(f"{num} x {multiply} = {num * multiply}")
```
