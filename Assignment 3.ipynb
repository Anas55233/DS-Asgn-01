```python
# Answer # 2

# Take input from user to check age of person
age = input("Enter Age: ")

# Now check a person belong to which age group by making conditions:
if age.isdigit():
    age = int(age)
    if age >= 0 and age <= 12:
        print(f"This Person is belong to 'Child' age group")
    elif age >= 13 and age <= 19:
        print(f"This Person is belong to 'Teenager' age group")
    elif age >= 20 and age <= 64:
        print(f"This Person is belong to 'Adult' age group")
    else:
        print(f"This Person is belong to 'Senior' age group")
else:
    print("Type Error!,Please Enter Positive digits Only ")

```

#### Answer No. 8

```python
# Answer No. 8

# Write Inventory
inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5},
    "Rich dad poor dad": {"price": 8.99, "quantity": 0},
    "Harry Potter": {"price": 12.50, "quantity": 2}
}

# Add New book in inventory and print updated inventory:
inventory.update({"HTML & CSS": {"price": 15, "quantity": 4}})
print(f"Updated Inventory: \n {inventory}")

# Get All book titles and print():
print(f"\nShow All Books Title: \n{inventory.items()}")

# Now Remove a book if it is out of stock.
for key, value in inventory.copy().items():
    if value["quantity"] == 0:
        inventory.pop(key)
        
print(f"\nInventory after removing out-of-stock books:\n{inventory}")
```
